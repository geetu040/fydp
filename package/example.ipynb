```python
from orator import connect_sql, describe_sql

db_paths = [
	"/home/geetu/work/orator/data/chinook/Chinook_Sqlite.sql",
]

db_conns = [connect_sql(i) for i in db_paths]
db_descs = [describe_sql(i) for i in db_conns]
```

```python
from orator import generate_sql

user_query = "many many columns and rows are in each table?"
response = generate_sql(user_query, db_descs[0])
```

```python
from orator import parse_response

sql_query = parse_response(response)
```

```python
from orator import execute_sql_query

data = execute_sql_query(sql_query, db_conns[0])
```

```python
from orator import create_data_response

response = create_data_response(user_query, data)
```
